## 1. Analysis of Loan Intent vs. Default Rates
   A.   OBJECTIVE - Identify which loan categories carry the highest credit risk for the lending institution.

   B. KEY FINDING:
 Debt Consolidation loans exhibit the highest average default rate across the entire portfolio, while Venture loans demonstrate the lowest risk profile.
 
   C. BUSINESS INSIGHT:
   Applicants borrowing capital to merge existing liabilities are statistically more vulnerable to cash-flow disruptions. Conversely, venture-backed loans generally indicate structured business modeling or higher baseline financial security.
   
   D.ACTIONABLE RECOMMENDATION:
   The risk department should implement stricter underwriting criteria or higher credit score thresholds specifically for applicants applying for debt consolidation.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Set the visual style
sns.set_theme(style="whitegrid")

# 2. Create the window size
plt.figure(figsize=(10, 6))

# 3. Draw the loan risk bars (X = loan purpose, Y = default rate)
# We added hue='loan_intent' and legend=False to match modern Seaborn standards
sns.barplot(x='loan_intent', y='loan_status', data=df, hue='loan_intent', palette='viridis', legend=False, errorbar=None)


# 4. Add clean, readable titles
plt.title("Loan Default Rates Across Different Intents", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Loan Purpose / Intent", fontsize=12)
plt.ylabel("Average Default Rate", fontsize=12)

# 5. Force the notebook to display the final graph
plt.show()
```

###  Evaluation of Interest Rate Distributions
     A. OBJECTIVE: Determine the statistical correlation between assigned loan interest rates and ultimate default outcomes.

     B. KEY FINDING: 
        The distribution for Defaulted Loans (Box 1) sits significantly higher on the vertical axis compared to Paid Back Loans (Box 0)
    
 
     C. BUSINESS INSIGHT: 
        Higher interest rates directly compound borrower strain. The elevated median interest rate for defaults proves that aggressive pricing tiers either attract lower-quality credit profiles or directly accelerate loan delinquency by inflating monthly debt-service obligations.

     D. STATISTICAL INSIGHT:
        The upper whisker and outliers in Box 1 underscore that loans priced above 12-14% carry an exponentially higher probability of asset impairment.


```python
# 1. Create a fresh canvas window
plt.figure(figsize=(8, 6))

# 2. Draw a box plot comparing loan status against interest rates
# X-axis: loan_status (0 = Paid Back, 1 = Defaulted)
# Y-axis: loan_int_rate (The interest rate percentage)
sns.boxplot(x='loan_status', y='loan_int_rate', data=df, hue='loan_status', palette='Set2', legend=False)

# 3. Customize the labels to look highly professional
plt.title("Distribution of Interest Rates by Loan Status", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Loan Status (0 = Paid Back | 1 = Defaulted)", fontsize=12)
plt.ylabel("Interest Rate (%)", fontsize=12)

# 4. Render the plot
plt.show()
```
